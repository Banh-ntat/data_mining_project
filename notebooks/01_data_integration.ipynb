# 01 - Kiểm tra và tích hợp dữ liệu

## Mục tiêu

- Đọc 4 bảng dữ liệu Olist.
- Kiểm tra cấu trúc dữ liệu.
- Kiểm tra dữ liệu thiếu và trùng.
- Xác định các khóa liên kết.
- Tích hợp các bảng thành một dataset.
- Xuất dữ liệu sau khi tích hợp.

```python
import pandas as pd
import numpy as np
```

```python
# đọc dữ liệu từ các file csv
orders = pd.read_csv('../data/raw/olist_orders_dataset.csv')
items = pd.read_csv('../data/raw/olist_order_items_dataset.csv')
customers = pd.read_csv('../data/raw/olist_customers_dataset.csv')
products = pd.read_csv('../data/raw/olist_products_dataset.csv')
```

```python
# kích thước của các bảng dữ liệu
print("Orders:", orders.shape)
print("Order items:", items.shape)
print("Customers:", customers.shape)
print("Products:", products.shape)
```

```python
# xem 5 dòng đầu tiên của các bảng dữ liệu
display(orders.head())
display(items.head())
display(customers.head())
display(products.head())
```

```python
# kiểm tra kiểu dữ liệu
orders.info()
print("\n")
items.info()
print("\n")
customers.info()
print("\n")
products.info()
```

```python
#kiểm tra dữ liệu thiếu
print("Orders:", orders.isna().sum())
print("\nItems:", items.isna().sum())
print("\nCustomers:", customers.isna().sum())
print("\nProducts:", products.isna().sum())
```

```python
# kiểm tra dữ liệu trùng lặp
print("Orders:", orders.duplicated().sum())
print("\nItems:", items.duplicated().sum())
print("\nCustomers:", customers.duplicated().sum())
print("\nProducts:", products.duplicated().sum())
```

## Xác định khóa liên kết

Các bảng được liên kết thông qua:

- `orders.customer_id` ↔ `customers.customer_id`
- `orders.order_id` ↔ `order_items.order_id`
- `order_items.product_id` ↔ `products.product_id`

```python
df = orders.merge(items, on='order_id', how='inner')
df = df.merge(customers, on='customer_id', how='inner')
df = df.merge(products, on='product_id', how='inner')
# in ra kich thước của bảng dữ liệu sau khi tích hợp
print("Integrated DataFrame:", df.shape)
```

```python
# kiểm tra 5 dòng đầu tiên của bảng dữ liệu sau khi tích hợp
display(df.head())
print("Số dòng:", df.shape[0])
print("Số cột:", df.shape[1])
df.info()
```

```python
#Kiểm tra xem merge có vấn đề không
print("Orders ban đầu:", len(orders))
print("Sau khi merge:", len(df))
```

```python
#Lưu kết quả tích hợp vào file csv
df.to_csv(
    '../data/processed/step1_merged.csv',
    index=False
)
```
